```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return (x + 3)**2

def grad_f(x):
    return 2 * (x + 3)
```

```python
x_current = 2
learning_rate = 0.1
tolerance = 1e-6
max_iterations = 100

history = [x_current]

print("Starting point:", x_current)
print("Initial function value:", f(x_current))
```

```python
for i in range(max_iterations):
    gradient = grad_f(x_current)

    x_next = x_current - learning_rate * gradient

    history.append(x_next)

    if abs(x_next - x_current) < tolerance:
        x_current = x_next
        print("Converged after", i + 1, "iterations.")
        break

    x_current = x_next

    print("Iteration", i + 1,
          ": x =", round(x_current, 6),
          ", y =", round(f(x_current), 6))
else:
    print("Maximum iterations reached.")
```

```python
print("Local minimum at x =", round(x_current, 6))
print("Minimum function value y =", round(f(x_current), 10))
```

```python
x_values = np.linspace(-5, 4, 100)
y_values = f(x_values)

plt.plot(x_values, y_values, label="y = (x + 3)^2")
plt.plot(history, [f(x) for x in history],
         "ro-", markersize=3, label="Gradient Descent")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent Convergence")
plt.legend()
plt.grid()
plt.show()
```
